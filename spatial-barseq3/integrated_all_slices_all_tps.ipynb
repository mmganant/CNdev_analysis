```python
import scanpy as sc
import matplotlib.pyplot as plt
import matplotlib.colors as mcolors


# Convert "Granule cells" → "Astroglia" for library_id containing e11 or e12
ct = adata_all.obs["cell_types"].astype("category")

mask = (
    adata_all.obs["library_id"].str.contains(r"e1[12]", case=False, na=False)
    & (ct.astype(str) == "Granule cells")
)

# ensure target category exists, then assign
if "Astroglia" not in ct.cat.categories:
    ct = ct.cat.add_categories(["Astroglia"])

ct = ct.where(~mask, other="Astroglia")

# (optional) drop unused categories
ct = ct.cat.remove_unused_categories()

adata_all.obs["cell_types"] = ct

# -------- 0) CONFIG --------
custom_colors = {
    "Purkinje Cells": "blue",
    "Interposed/Lateral DCN": "red",
    "Medial DCN": "orange",
    "DCN": "orange",
    'Astroglia': 'grey',
    "": "white",
    "Choroid Plexus": "black",
    "Granule cells": "green",
    "Midbrain-fated cells": "brown",
    "Molecular Layer Interneurons": "pink",
    "Outside Cb": "purple",
    'Glia/Oligodendrocytes': "lightblue",
    "i1": "yellow",
}

# Drawn from bottom to top (last wins). Put "Medial DCN" last if you want it on top.
desired_order = [
    "Outside Cb",
    "Choroid Plexus",
    'Astroglia',
    'Glia/Oligodendrocytes',
    "Molecular Layer Interneurons",
    "Granule cells",
    "DCN",
    "Purkinje Cells",
    "",  # empty-label if present
    "Interposed/Lateral DCN",
    "Medial DCN",    
    "Midbrain-fated cells",# <- plotted last (on top)
    "i1",
]

to_plot = [
    ("e11_1",  80,  "all_tp_e11"),
    ("e12_1",  80,  "all_tp_e12"),
    ("e13_5",  80,  "all_tp_e13"),
    ("e14_2", 100,  "all_tp_e14"),
    ("e15_2", 120,  "all_tp_e15"),
    ("e17_1", 120,  "all_tp_e17"),
    ("p0_1",  120,  "all_tp_p0"),
    ("p4_2",  120,  "all_tp_p4"),
    ("p56_1", 200,  "all_tp_p56"),
]

# -------- 1) COLOR + ORDER HELPERS --------
def set_celltype_colors(adata, color_map):
    adata.obs["cell_types"] = adata.obs["cell_types"].astype("category")
    cats = list(adata.obs["cell_types"].cat.categories)
    missing = [c for c in cats if c not in color_map]
    if missing:
        raise ValueError(f"color_map missing categories: {missing}")
    adata.uns["cell_types_colors"] = [mcolors.to_hex(color_map[c]) for c in cats]

def set_celltype_order(adata, desired_order):
    ct = adata.obs["cell_types"].astype("category")
    cur = list(ct.cat.categories)
    final_order = [c for c in desired_order if c in cur] + [c for c in cur if c not in desired_order]
    # reorder colors accordingly
    color_by_cat = dict(zip(cur, adata.uns["cell_types_colors"]))
    adata.uns["cell_types_colors"] = [color_by_cat[c] for c in final_order]
    # apply order
    adata.obs["cell_types"] = ct.cat.reorder_categories(final_order, ordered=True)

# -------- 2) OVERLAY PLOTTER (avoids grey repaint) --------
def spatial_overlay_in_order(
    ad, color="cell_types", order=None, spot_size=80,
    legend_loc="right margin", save=None, show=True
):
    ad.obs[color] = ad.obs[color].astype("category")

    cats_present = [c for c in ad.obs[color].cat.categories if (ad.obs[color] == c).any()]
    draw_order = [c for c in (order or cats_present) if c in cats_present]
    if not draw_order:
        print("[warn] no matching categories in this slice; drawing base only")
        sc.pl.spatial(ad, color=color, spot_size=spot_size, show=show)
        return

    # base: draw all spots (light NA so structure is visible)
    sc.pl.spatial(
        ad, color=color, spot_size=spot_size,
        na_color="#DDDDDD", legend_loc=None, show=False
    )
    ax = plt.gca()

    # overlays: only current group, others transparent
    for g in draw_order:
        sc.pl.spatial(
            ad, color=color, groups=[g], spot_size=spot_size,
            na_color="none", ax=ax, legend_loc=None, title=None, show=False
        )

    # one legend pass (no points)
    sc.pl.spatial(
        ad, color=color, spot_size=0, ax=ax, legend_loc=legend_loc,
        na_in_legend=False, show=False
    )

    if save:
        import matplotlib as mpl

        mpl.rcParams.update({
            "font.family": "sans-serif",
            "font.sans-serif": ["Arial"],
            "svg.fonttype": "none",   # keeps text editable in Illustrator
            "pdf.fonttype": 42,
        })


        plt.savefig(
            save,
            bbox_inches="tight"
        )
    if show:
        plt.show()
    plt.close()

# -------- 3) APPLY ON THE COMBINED OBJECT --------
# set colors & order once on the full object
set_celltype_colors(adata_all, custom_colors)
set_celltype_order(adata_all, desired_order)

sc.set_figure_params(figsize=(15, 5))

# -------- 4) DRAW EACH LIBRARY WITH SAME COLORS/ORDER --------
for lib_id, spot, suffix in to_plot:
    ad = adata_all[adata_all.obs["library_id"] == lib_id].copy()
    if ad.n_obs == 0:
        print(f"[warn] No cells for library_id={lib_id}, skipping")
        continue
    # keep the global categories & colors
    ad.obs["cell_types"] = ad.obs["cell_types"].cat.set_categories(
        adata_all.obs["cell_types"].cat.categories, ordered=True
    )
    ad.uns["cell_types_colors"] = adata_all.uns["cell_types_colors"]

    spatial_overlay_in_order(
        ad, color="cell_types", order=desired_order,
        spot_size=spot, show=True,
        save=f"{suffix}_overlay.svg",
    )
```
